# NIFTY Candlestick Analyzer - Exploratory Analysis

This notebook demonstrates the usage of nfscan package for analyzing NIFTY candlestick patterns.

```python
# Import required modules
import sys
sys.path.insert(0, '..')

from nfscan import DataFetcher, PatternDetector, Visualizer
import pandas as pd
import matplotlib.pyplot as plt

# Display settings
pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', 20)

print("✅ Imports successful!")
```

## 1. Fetch NIFTY Data

```python
# Create data fetcher instance
fetcher = DataFetcher("^NSEI")  # NIFTY 50

# Fetch 6 months of daily data
data = fetcher.fetch_data(period="6mo", interval="1d")

print(f"Fetched {len(data)} data points")
data.tail()
```

## 2. Basic Statistics

```python
# Display basic statistics
print("Basic Statistics:")
print(f"Latest Close: ₹{data['Close'].iloc[-1]:.2f}")
print(f"Period High: ₹{data['High'].max():.2f}")
print(f"Period Low: ₹{data['Low'].min():.2f}")
print(f"Average Volume: {data['Volume'].mean():.0f}")

data.describe()
```

## 3. Detect Candlestick Patterns

```python
# Create pattern detector
detector = PatternDetector(data)

# Detect all patterns
patterns = detector.detect_all_patterns()

# Display pattern summary
summary = detector.get_pattern_summary()
summary
```

## 4. Visualize Patterns

```python
# Create visualizer
visualizer = Visualizer(data)

# Plot candlestick chart
fig = visualizer.plot_candlestick(title="NIFTY 50 Candlestick Chart")
plt.show()
```

```python
# Plot candlestick chart with patterns
fig = visualizer.plot_with_patterns(patterns, title="NIFTY 50 with Detected Patterns")
plt.show()
```

```python
# Plot price movement
fig = visualizer.plot_price_movement(title="NIFTY 50 Price Movement")
plt.show()
```

```python
# Plot pattern distribution
fig = visualizer.plot_pattern_distribution(summary, title="Pattern Distribution")
plt.show()
```

## 5. Analyze Specific Patterns

```python
# Get dates where Bullish Engulfing pattern occurred
try:
    bullish_dates = detector.get_pattern_dates('Bullish Engulfing')
    print(f"Bullish Engulfing pattern detected on {len(bullish_dates)} days:")
    for date in bullish_dates[-5:]:  # Show last 5
        print(f"  {date.strftime('%Y-%m-%d')}: Close = ₹{data.loc[date, 'Close']:.2f}")
except Exception as e:
    print(f"Error: {e}")
```

```python
# Get dates where Hammer pattern occurred
try:
    hammer_dates = detector.get_pattern_dates('Hammer')
    print(f"Hammer pattern detected on {len(hammer_dates)} days:")
    for date in hammer_dates[-5:]:  # Show last 5
        print(f"  {date.strftime('%Y-%m-%d')}: Close = ₹{data.loc[date, 'Close']:.2f}")
except Exception as e:
    print(f"Error: {e}")
```

## 6. Export Results

```python
# Save data to CSV
fetcher.save_data('../data/nifty_data.csv')
print("✅ Data saved to data/nifty_data.csv")

# Save pattern summary
summary.to_csv('../data/pattern_summary.csv', index=False)
print("✅ Pattern summary saved to data/pattern_summary.csv")
```
